```python
%matplotlib widget
import matplotlib.pyplot as plt
import numpy as np
from ipywidgets import interact, FloatSlider

def plot_sine(frequency=1.0):
    """Plot sine wave with adjustable frequency"""
    x = np.linspace(0, 4*np.pi, 500)
    y = np.sin(frequency * x)
    
    plt.figure(figsize=(10, 4))
    plt.plot(x, y, linewidth=2)
    plt.grid(True, alpha=0.3)
    plt.title(f'Sine Wave with Frequency = {frequency:.2f}')
    plt.xlabel('x')
    plt.ylabel('sin(frequency × x)')
    plt.ylim(-1.5, 1.5)
    plt.show()

# Create interactive slider
interact(plot_sine, 
         frequency=FloatSlider(min=0.5, max=5.0, step=0.1, value=1.0, 
                              description='Frequency:'));
```
